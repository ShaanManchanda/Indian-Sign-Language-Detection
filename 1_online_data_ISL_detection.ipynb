# 2. Keypoints using MP Holistic

```python
# Initialize MediaPipe Hands.
def initialize_hand_tracking():
    """Initializes and returns the MediaPipe Hands object."""
    mp_hands = mp.solutions.hands
    hands = mp_hands.Hands(
        static_image_mode=False,
        max_num_hands=2,
        min_detection_confidence=0.5,
        min_tracking_confidence=0.5)
    return hands, mp_hands
```

```python
# Initialize MediaPipe Hands object
def detect_hands(image, hands):
    """Processes the image and detects hands using the initialized MediaPipe Hands object."""
    # Convert the BGR image to RGB.
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    # To improve performance, mark the image as not writeable.
    image_rgb.flags.writeable = False
    # Process the image and detect hands.
    results = hands.process(image_rgb)
    # Convert back to BGR for display and writing.
    image_rgb.flags.writeable = True
    image_bgr = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2BGR)
    return image_bgr, results
```

```python
# Initialize MediaPipe drawing module.
def draw_hand_landmarks(image, results, mp_hands):
    """Draws hand landmarks on the image."""
    if results.multi_hand_landmarks:
        for hand_landmarks in results.multi_hand_landmarks:
            mp.solutions.drawing_utils.draw_landmarks(
                image,
                hand_landmarks,
                mp_hands.HAND_CONNECTIONS,
                mp.solutions.drawing_utils.DrawingSpec(color=(121, 22, 76), thickness=2, circle_radius=4),
                mp.solutions.drawing_utils.DrawingSpec(color=(250, 44, 250), thickness=2))
    return image
```

```python
def process_webcam():
    """Processes video input from the webcam to detect and draw hand landmarks."""
    hands, mp_hands = initialize_hand_tracking()
    cap = cv2.VideoCapture(0)  # Capture video from the webcam

    if not cap.isOpened():
        print("Failed to open webcam.")
        return

    while cap.isOpened():
        success, image = cap.read()
        if not success:
            print("Ignoring empty camera frame.")
            continue

        # Detect hands and process the image.
        image, results = detect_hands(image, hands)
        # Draw hand landmarks on the image.
        image = draw_hand_landmarks(image, results, mp_hands)

        # Flip the image horizontally for a selfie-view display and show it.
        cv2.imshow('MediaPipe Hands', cv2.flip(image, 1))

        # Press 'q' to exit the loop.
        if cv2.waitKey(5) & 0xFF == ord('q'):
            break

    # Release the webcam and close the window.
    cap.release()
    cv2.destroyAllWindows()
```

```python
# write mediapipe landmarks to video and save to video
def process_video(input_path, output_path):
    """Processes a video file to detect and draw hand landmarks."""
    hands, mp_hands = initialize_hand_tracking()
    cap = cv2.VideoCapture(input_path)
    if not cap.isOpened():
        print("Error: Unable to open video file.")
        return

    frame_width, frame_height = int(cap.get(3)), int(cap.get(4))
    frame_rate = cap.get(cv2.CAP_PROP_FPS)
    fourcc = cv2.VideoWriter_fourcc(*'XVID')
    out = cv2.VideoWriter(output_path, fourcc, frame_rate, (frame_width, frame_height))

    while cap.isOpened():
        success, image = cap.read()
        if not success:
            break
        
        # Correctly unpack the image and results
        image, results = detect_hands(image, hands)
        image = draw_hand_landmarks(image, results, mp_hands)
        out.write(image)

    cap.release()
    out.release()
    cv2.destroyAllWindows()
    print("Video processing completed. Output saved as:", output_path)
```

# 3. Extract Keypoint Values

```python
# Extracting Keypoint Values from a video file
def extract_hand_landmarks(input_path):
    """Extracts hand landmarks from a video file."""
    hands, mp_hands = initialize_hand_tracking()
    cap = cv2.VideoCapture(input_path)
    all_landmarks = []

    if not cap.isOpened():
        print("Error: Unable to open video file.")
        return None

    while cap.isOpened():
        success, image = cap.read()
        if not success:
            break
        
        image, results = detect_hands(image, hands)
        if results.multi_hand_landmarks:
            frame_landmarks = []
            for hand_landmarks in results.multi_hand_landmarks:
                # Extract x, y, z coordinates of landmarks
                landmarks = [(lm.x, lm.y, lm.z) for lm in hand_landmarks.landmark]
                frame_landmarks = frame_landmarks + landmarks
            all_landmarks.append(frame_landmarks)
    
    cap.release()
    return all_landmarks
```

### 1 - Normalize the Data

Normalization helps to scale the data so that the model trains more efficiently and effectively. Given that hand landmark coordinates from MediaPipe are typically normalized between 0 and 1 relative to the dimensions of the image frame, you might only need to ensure this is consistent across all data. If additional scaling is needed, here’s how you can normalize:

```python
import numpy as np

def normalize_landmarks(data):
    normalized_data = []
    for video in data:
        normalized_video = []
        for frame in video:
            normalized_frame = []
            for hand_landmarks in frame:
                # Convert landmarks to numpy array for easier manipulation
                landmarks_array = np.array(hand_landmarks)
                
                # Example normalization: standardize landmarks (if needed, not typically required)
                # landmarks_centered = landmarks_array - np.mean(landmarks_array, axis=0)
                # normalized_frame.append(landmarks_centered)

                # Since MediaPipe landmarks are already normalized between 0 and 1,
                # you might not need further normalization, so we append them directly.
                normalized_frame.append(landmarks_array)
            normalized_video.append(normalized_frame)
        normalized_data.append(normalized_video)
    return normalized_data
```

### Flatten Landmarks per Frame

```python
def flatten_landmarks(data):
    flattened_data = []
    for video in data:
        video_frames = []
        for frame in video:
            # Flatten frame landmarks from all hands into a single list
            frame_landmarks = []
            for hand in frame:
                for lm in hand:
                    frame_landmarks.extend(lm)
            video_frames.append(frame_landmarks)
        flattened_data.append(video_frames)
    return flattened_data
```

### 2 - Pad Sequences

To ensure that all input sequences are of the same length, use padding. This is crucial for effective training of LSTMs, which require fixed-length inputs.

```python
def pad_landmark_sequences(landmarks, maxlen=100):
    padded_data = []
    for video in landmarks:
        # Pad video frames so all videos have the same number of frames, and each frame has the same size
        video_padded = pad_sequences(video, maxlen=maxlen, dtype='float32', padding='post', truncating='post', value=0.0)
        padded_data.append(video_padded)
    return np.array(padded_data, dtype=object)  # Use dtype=object if arrays have different shapes
```

```python
from sklearn.model_selection import train_test_split

def split_data(landmarks, labels, test_size=0.2, random_state=42):
    X_train, X_test, y_train, y_test = train_test_split(landmarks, labels, test_size=test_size, random_state=random_state)
    return X_train, X_test, y_train, y_test
```
